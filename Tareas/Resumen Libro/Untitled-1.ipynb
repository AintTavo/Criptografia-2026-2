## Tipos de crifrados

Los desarrollos realizados durante todos estos años en el area han dado como resultado dos corrientes para realizar un cifrado: Transposición y Sustitución

### Cifrado de transposición

En este tomamos el mensaje e intercambiamos la posición de las letras del mensaje en si, no se añade ninguna sino cambian posiciones entre estas mismas, siendo la forma de encriptarlo y desencriptarlo saber el orden en que fueron acomodadas. Esto resulta extremadamente sencillo de transpasar para mensajes extremadamente cortos como podria ser la palabra:

aacv => vaca

Justamente las formas posibles para responder esto por fuerza bruta al elevarse de manera factorial se vuelve inviable un ataque de fuerza bruta porque en este caso de 5 seria:

4! = 24

pero para una mensaje de simplemente 20 caracteres seria:

20! = 2.43x10¹⁸

Si probaramos cada combinación cada milisegundo tardariamos aproximadamente 77 millones de años en probar todas.

### Cifrado de sustitución

El cifrado de sustitución funciona relativamente diferente al cifrado de sustitución y este es de que todos los simbolos de un texto pasan a ser otro dentro de un alfabeto definido, un ejemplo seria de que la 'a' pasa a ser la 'D' esto no suena tan dificil a simple vista, principalmente porque conocemos que una letra pasa a ser otra, pero la naturaleza de este tipo de cifrado es exponencial y modular. El modulo no solo define los los limetes de simbolos del sistema, pudiendo aplicarles funciones a estos para determinar su sustitución, sino tambien la cantidad posible de resultados escala de manera drastica; para un abecedario con 27 letras como el español y con un mensaje de 20, este puede representar el siguiente numero de mensajes por sustitución

27²⁰ = 4.23x10²⁸
